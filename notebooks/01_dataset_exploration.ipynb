```python
from pathlib import Path

dataset_path = Path("../datasets/FakeAVCeleb_v1.2/FakeAVCeleb_v1.2")

classes = [
    "FakeVideo-FakeAudio",
    "FakeVideo-RealAudio",
    "RealVideo-FakeAudio",
    "RealVideo-RealAudio"
]

for cls in classes:
    folder = dataset_path / cls

    videos = list(folder.rglob("*.mp4"))

    print(f"{cls}: {len(videos)} videos")
```
